## Random Tensor

why random tensor?

Random tensors are important because the way many neural networks learn is that they start with tensors full of random numbers then adjust those numbers to better represent the data.

```python
## Random Tensor
## creating a random tensor of size 3X4
random_tensor = torch.rand(3,4)
random_tensor
```

```python
random_tensor.ndim
```

```python
# create a random tensor with similar shape to an image tensor
random_image_size_tensor = torch.rand(size=(3,4)) # height, width, color channels (R, G, B)
random_image_size_tensor.shape, random_image_size_tensor.ndim
```

```python
torch.rand(size=(3,3))
```

## Tensor Datatypes

*NOTE :* Tensors datatypes is one of the 3 big errors we'll run into pytorch & deep learning
1. Tensors not right Datatype
2. Tensors not right shape
3. Tensors not on the right device

```python
# Float 32 tensor
float_32_tensor = torch.tensor([3.0, 6.0, 9.0],
                               dtype=None, # what datatype is a tensor(eg. float)
                               device=None,# whar device is your tensor on
                               requires_grad=False) # wheather or not to track gradeints with this tensor operation
float_32_tensor
```

```python
float_32_tensor.dtype
```

```python
float_16_tensor = float_32_tensor.type(torch.float16)
float_16_tensor
```

```python
float_32_tensor * float_16_tensor
```

```python
int_32_tensor = torch.tensor([3,4,9],dtype=torch.int32)
int_32_tensor
```

```python
float_32_tensor * int_32_tensor
```

### Getting information from tensor

1. Tensors not right Datatype - to get datatype from a tensor, can use tensor.dtype
2. Tensors not right shape- to get shape from a tensor, can use tensor.shape
3. Tensors not on the right device- to get a device from a tensor, can use tensor.device



```python
# Create a tensor

some_tensor = torch.rand(3,4)
some_tensor
```

```python
# find out details about some tensor
print(some_tensor)
print(f"Datatype of tensor: {some_tensor.dtype}")
print(f"Shape of tensor: {some_tensor.shape}")
print(f"Device tensor is on: {some_tensor.device}")
```

## Matrix Multiplication

Two main ways of performing Multiplication in nural network and deep learning

1.Element-wise Multiplication
2.Matrix multiplication(dot product)


```python
# Element wise multiplication
print(tensor, "*", tensor)
print(f"Equals: {tensor * tensor}")
```

```python
# Matrix multiplication
torch.matmul(tensor,tensor)
```
