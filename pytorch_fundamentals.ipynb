## Random Tensor

why random tensor?

Random tensors are important because the way many neural networks learn is that they start with tensors full of random numbers then adjust those numbers to better represent the data.

```python
## Random Tensor
## creating a random tensor of size 3X4
random_tensor = torch.rand(3,4)
random_tensor
```

```python
random_tensor.ndim
```

```python
# create a random tensor with similar shape to an image tensor
random_image_size_tensor = torch.rand(size=(3,4)) # height, width, color channels (R, G, B)
random_image_size_tensor.shape, random_image_size_tensor.ndim
```

```python
torch.rand(size=(3,3))
```

```python
torch.rand(3,3)
```

### Creating a range of tensor and tensor-like


```python
# Use torch.range() and get deprecated message, use torch.arange()
one_to_ten=torch.arange(start=0, end=11,step=1)
one_to_ten
```

```python
# Creating Tensor Like
one_to_ten_like = torch.zeros_like(input=one_to_ten)
one_to_ten_like
```
